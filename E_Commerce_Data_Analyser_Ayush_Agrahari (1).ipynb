## Step 4: Clean the data
1. Find missing values
2. Find duplicate rows
3. Fix both

```python
print("Missing values in each column:")
print(df.isnull().sum())

print()
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Fill the missing rating with the average rating
avg_rating = df["rating"].mean()
df["rating"] = df["rating"].fillna(avg_rating)

print("Rows after cleaning:", len(df))
print("Missing values left:", df.isnull().sum().sum())
print("Duplicate rows left:", df.duplicated().sum())
```

## Step 5: Make a new column
**total_amount = price × quantity**

```python
df["total_amount"] = df["price"] * df["quantity"]

df[["order_id", "product", "price", "quantity", "total_amount"]].head()
```

## Step 6: Main business numbers

```python
print("Total orders        :", len(df))
print("Total customers     :", df["customer"].nunique())
print("Total items sold    :", df["quantity"].sum())
print("Total revenue (Rs.) :", df["total_amount"].sum())
print("Average order (Rs.) :", round(df["total_amount"].mean(), 2))
print("Biggest order (Rs.) :", df["total_amount"].max())
print("Smallest order (Rs.):", df["total_amount"].min())
print("Average rating      :", round(df["rating"].mean(), 2))
```

## Step 7: Statistics using NumPy
We convert the `total_amount` column into a NumPy array and use NumPy functions.

```python
amounts = np.array(df["total_amount"])

print("Sum              :", np.sum(amounts))
print("Mean (average)   :", round(np.mean(amounts), 2))
print("Median (middle)  :", np.median(amounts))
print("Maximum          :", np.max(amounts))
print("Minimum          :", np.min(amounts))
print("Standard deviation:", round(np.std(amounts), 2))
```

## Step 8: Find answers using `groupby`
`groupby` puts rows with the same value together (for example, all rows of the same category) and then we add them up.

### 8.1 Revenue of each category

```python
category_sales = df.groupby("category")["total_amount"].sum().sort_values(ascending=False)
print(category_sales)
```

### 8.2 Revenue of each product

```python
product_sales = df.groupby("product")["total_amount"].sum().sort_values(ascending=False)
print(product_sales)
```

### 8.3 Revenue of each month

```python
month_sales = df.groupby("month")["total_amount"].sum()

# put months in the correct order
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
month_sales = month_sales[month_order]
print(month_sales)
```

### 8.4 Revenue of each city

```python
city_sales = df.groupby("city")["total_amount"].sum().sort_values(ascending=False)
print(city_sales)
```

### 8.5 Which payment method is used most?

```python
payment_count = df["payment"].value_counts()
print(payment_count)
```

### 8.6 Average rating of each category

```python
rating_by_category = df.groupby("category")["rating"].mean().round(2)
print(rating_by_category)
```

### 8.7 Top 3 customers (who spent the most)

```python
top_customers = df.groupby("customer")["total_amount"].sum().sort_values(ascending=False).head(3)
print(top_customers)
```

## Step 9: Final result

```python
print("======= FINAL RESULT =======")
print("Total revenue            : Rs.", df["total_amount"].sum())
print("Best category            :", category_sales.idxmax())
print("Best product             :", product_sales.idxmax())
print("Best month               :", month_sales.idxmax())
print("Best city                :", city_sales.idxmax())
print("Most used payment method :", payment_count.idxmax())
print("Best rated category      :", rating_by_category.idxmax())
```
